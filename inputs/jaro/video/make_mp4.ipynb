```python
import cv2
import glob
import os

# Folder where your images are stored
image_folder = "../"

# Camera prefix
camera_name = "gopro13"

# Pattern to match files from this camera
pattern = os.path.join(image_folder, f"{camera_name}_*.jpg")

# Collect and sort all matching images
images = sorted(glob.glob(pattern))

if not images:
    raise FileNotFoundError(f"No images found matching {pattern}")

# Read first image to get video size
frame = cv2.imread(images[0])
height, width, _ = frame.shape

# Set output video name to camera name
output_name = f"{camera_name}.mp4"

# Create video writer at 60 fps
fps = 60
out = cv2.VideoWriter(output_name, cv2.VideoWriter_fourcc(*"mp4v"), fps, (width, height))

# Write each image for 1 second (60 frames)
for img_path in images:
    frame = cv2.imread(img_path)
    out.write(frame)

out.release()
print(f"Video written: {output_name} with {len(images)} images at {fps}fps")
```
