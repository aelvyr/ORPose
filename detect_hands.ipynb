```python
import cv2
from cvzone.HandTrackingModule import HandDetector

cap = cv2.VideoCapture("inputs/gopro7_synced_cut.MP4")

hand_detector = HandDetector(detectionCon=0.05)
success = True
while success:
    success, img = cap.read()
    if not success:
        break
    list_hands, img = hand_detector.findHands(img)
    print(list_hands)
    cv2.namedWindow('image', cv2.WINDOW_NORMAL)
    cv2.resizeWindow('image', 1920, 1080)
    cv2.imshow('image', img)
    if cv2.waitKey(1) == 113:
        success = False

cv2.destroyAllWindows()
```
